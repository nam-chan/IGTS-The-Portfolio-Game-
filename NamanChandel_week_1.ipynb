```python
def iesds(matrix, row_labels=None, col_labels=None):
    rows = len(matrix)
    cols = len(matrix[0])
    if row_labels is None:
        row_labels = [f"R{i}" for i in range(rows)]
    if col_labels is None:
        col_labels = [f"C{i}" for i in range(cols)]
    active_rows = list(range(rows))
    active_cols = list(range(cols))
    changed = True
    while changed:
        changed = False
        row_to_remove = None
        for r1 in active_rows:
            for r2 in active_rows:
                if r1 == r2:
                    continue
                dominated = True
                for c in active_cols:
                    payoff1 = matrix[r1][c][0]
                    payoff2 = matrix[r2][c][0]
                    if payoff1 >= payoff2:
                        dominated = False
                        break
                if dominated:
                    row_to_remove = r1
                    print(
                        f"Remove Row {row_labels[r1]} "
                        f"(strictly dominated by Row {row_labels[r2]})"
                    )

                    break

            if row_to_remove is not None:
                break
        if row_to_remove is not None:
            active_rows.remove(row_to_remove)
            changed = True
            continue
        col_to_remove = None
        for c1 in active_cols:
            for c2 in active_cols:
                if c1 == c2:
                    continue
                dominated = True
                for r in active_rows:
                    payoff1 = matrix[r][c1][1]
                    payoff2 = matrix[r][c2][1]
                    if payoff1 >= payoff2:
                        dominated = False
                        break
                if dominated:
                    col_to_remove = c1

                    print(
                        f"Remove Column {col_labels[c1]} "
                        f"(strictly dominated by Column {col_labels[c2]})"
                    )
                    break
            if col_to_remove is not None:
                break
        if col_to_remove is not None:
            active_cols.remove(col_to_remove)
            changed = True
    print("\nRemaining Rows:")
    for r in active_rows:
        print(row_labels[r])
    print("\nRemaining Columns:")
    for c in active_cols:
        print(col_labels[c])
    if len(active_rows) == 1 and len(active_cols) == 1:

        r = active_rows[0]
        c = active_cols[0]

        print("\nPredicted Outcome:")
        print(
            f"({row_labels[r]}, {col_labels[c]}) "
            f"with payoff {matrix[r][c]}"
        )

matrix = [

    [(4,3),  (2,2),  (3,1),  (1,0)],
    [(5,4),  (3,3),  (4,2),  (2,1)],
    [(1,5),  (0,4),  (2,3),  (1,2)]
]

row_labels = ["A", "B", "C"]
col_labels = ["D", "E", "F", "G"]
iesds(matrix, row_labels, col_labels)
```

```python
def find_nash_equilibria(matrix, row_labels=None, col_labels=None):
    rows = len(matrix)
    cols = len(matrix[0])

    if row_labels is None:
        row_labels = [f"R{i}" for i in range(rows)]

    if col_labels is None:
        col_labels = [f"C{i}" for i in range(cols)]

    p1_best = set()
    p2_best = set()

    for c in range(cols):
        max_payoff = max(matrix[r][c][0] for r in range(rows))

        for r in range(rows):
            if matrix[r][c][0] == max_payoff:
                p1_best.add((r, c))
    for r in range(rows):
        max_payoff = max(matrix[r][c][1] for c in range(cols))
        for c in range(cols):
            if matrix[r][c][1] == max_payoff:
                p2_best.add((r, c))
    equilibria = p1_best.intersection(p2_best)
    print("Game Matrix:\n")
    for r in range(rows):
        print(row_labels[r], end=" : ")
        for c in range(cols):
            print(matrix[r][c], end="  ")
        print()
    print("\nP1 Best Responses:")
    for cell in sorted(p1_best):
        r, c = cell
        print(f"({row_labels[r]}, {col_labels[c]})")
    print("\nP2 Best Responses:")
    for cell in sorted(p2_best):
        r, c = cell
        print(f"({row_labels[r]}, {col_labels[c]})")
    print("\nPure Strategy Nash Equilibria:")
    if equilibria:
        for r, c in sorted(equilibria):
            print(
                f"({row_labels[r]}, {col_labels[c]}) "
                f"with payoff {matrix[r][c]}"
            )

    else:
        print("No pure-strategy Nash equilibrium found.")

    print("\n" + "=" * 60 + "\n")

prisoners_dilemma = [
    [(-1,-1), (-5, 0)],
    [( 0,-5), (-3,-3)]
]

print("PRISONER'S DILEMMA")

find_nash_equilibria(
    prisoners_dilemma,
    ["Cooperate", "Defect"],
    ["Cooperate", "Defect"]
)

stag_hunt = [
    [(4,4), (1,3)],
    [(3,1), (2,2)]
]

print("STAG HUNT")

find_nash_equilibria(
    stag_hunt,
    ["Stag", "Hare"],
    ["Stag", "Hare"]
)
chicken = [
    [(0,0), (-1,1)],
    [(1,-1), (-10,-10)]
]
print("CHICKEN")
find_nash_equilibria(
    chicken,
    ["Swerve", "Straight"],
    ["Swerve", "Straight"]
)
battle = [
    [(2,1), (0,0)],
    [(0,0), (1,2)]
]
print("BATTLE OF THE SEXES")
find_nash_equilibria(
    battle,
    ["Opera", "Football"],
    ["Opera", "Football"]
)
matrix_3x3 = [
    # X         Y         Z
    [(3,2),   (1,1),   (0,0)],   # A
    [(2,3),   (4,2),   (1,1)],   # B
    [(0,0),   (2,4),   (5,5)]    # C
]
print("3x3 MATRIX")
find_nash_equilibria(
    matrix_3x3,
    ["A", "B", "C"],
    ["X", "Y", "Z"]
)
```

```python
from itertools import product
def nash_equilibria(strategies, payoffs):
    players = len(strategies)
    equilibria = []
    profiles = list(product(*strategies))
    for profile in profiles:
        nash = True
        for player in range(players):
            current_payoff = payoffs[profile][player]

            for alt_strategy in strategies[player]:
                if alt_strategy == profile[player]:
                    continue


                new_profile = list(profile)
                new_profile[player] = alt_strategy
                new_profile = tuple(new_profile)


                if payoffs[new_profile][player] > current_payoff:
                    nash = False
                    break

            if not nash:
                break

        if nash:
            equilibria.append(profile)

    return equilibria
strategies = [['C','D'], ['C','D'], ['C','D']]

payoffs = {
    ('C','C','C'): [4, 4, 4],
    ('C','C','D'): [1, 1, 6],
    ('C','D','C'): [1, 6, 1],
    ('C','D','D'): [0, 3, 3],
    ('D','C','C'): [6, 1, 1],
    ('D','C','D'): [3, 0, 3],
    ('D','D','C'): [3, 3, 0],
    ('D','D','D'): [2, 2, 2]
}

equilibria = nash_equilibria(strategies, payoffs)

if equilibria:
    print("Pure-strategy Nash Equilibria:")
    for eq in equilibria:
        print(eq)
else:
    print("No pure-strategy Nash equilibria found.")
```
