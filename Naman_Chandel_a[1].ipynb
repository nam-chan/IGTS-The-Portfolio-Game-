# Q7

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(67)
returns = np.random.normal(loc=0.0008, scale=0.015, size=252)
prices = 100 * np.cumprod(1 + returns)
prices = np.insert(prices, 0, 100)
plt.figure(figsize=(10, 6))
plt.plot(prices, color='blue', label='Simulated Price')
plt.title('Simulated Asset Price Series (252 Days)')
plt.xlabel('Trading Days')
plt.ylabel('Price')
plt.grid(True)
plt.tight_layout()
plt.show()
final_price = prices[-1]
peak_price = np.max(prices)
running_max = np.maximum.accumulate(prices)
drawdowns = (running_max - prices) / running_max
max_drawdown = np.max(drawdowns) * 100
print(f"Final Price: {final_price:.2f}")
print(f"Peak Price: {peak_price:.2f}")
print(f"Maximum Drawdown: {max_drawdown:.2%}")
```

# Q8

```python
import pandas as pd
import matplotlib.pyplot as plt
months = ['Jan','Feb','Mar','Apr','May','Jun',
          'Jul','Aug','Sep','Oct','Nov','Dec']

year1 = [42, 45, 50, 48, 55, 60, 58, 63, 67, 70, 75, 80]

year2 = [52, 49, 58, 61, 66, 71, 68, 74, 79, 83, 88, 95]
df = pd.DataFrame({
    'Month': months,
    'Year1': year1,
    'Year2': year2
})
df['Difference'] = df['Year2'] - df['Year1']
colors = ['green' if x > 0 else 'red' for x in df['Difference']]
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))
df.plot(x='Month', y=['Year1', 'Year2'],
        kind='line', marker='o', ax=ax1)

ax1.set_title('Monthly Revenue Comparison')
ax1.set_xlabel('Months')
ax1.set_ylabel('Revenue (in lakhs)')
ax1.grid(True)
ax2.bar(df['Month'], df['Difference'], color=colors)

ax2.set_title('Difference Between Year2 and Year1 Revenue')
ax2.set_xlabel('Months')
ax2.set_ylabel('Difference')
plt.tight_layout()
plt.show()
```

# Q10

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
returns = np.random.multivariate_normal(
mean=[0.0005, 0.0008, 0.0003],
cov=[[0.0004, 0.0001, 0.00005],
[0.0001, 0.0006, 0.00008],
[0.00005, 0.00008, 0.0003]],
size=200
)
prices = 100 * np.cumprod(1 + returns, axis=0)
print(np)
df = pd.DataFrame(prices, columns=['Asset_A', 'Asset_B', 'Asset_C'])
df_returns = pd.DataFrame(returns, columns=['Asset_A', 'Asset_B', 'Asset_C'])
corr_matrix = df_returns.corr()
print("Correlation Matrix of Daily Returns:")
print(corr_matrix.round(4))
plt.figure(figsize=(10, 6))
portfolio_returns = df_returns.mean(axis=1)
df['Equally_Weighted_Portfolio'] = 100 * np.cumprod(1 + portfolio_returns)
plt.plot(df['Asset_A'], label='Asset A', color='blue', alpha=0.67)
plt.plot(df['Asset_B'], label='Asset B', color='green', alpha=0.67)
plt.plot(df['Asset_C'], label='Asset C', color='orange', alpha=0.67)
plt.plot(df['Equally_Weighted_Portfolio'], label='Equally-Weighted Portfolio',
         color='black', linewidth=2, linestyle='--')
plt.title('Simulated Asset Prices and Equally-Weighted Portfolio (200 Days)')
plt.xlabel('Trading Days')
plt.ylabel('Price')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
